```python
reading=np.genfromtxt("StudentsPerformance.csv",delimiter=',',skip_header=1,dtype=str,usecols=6)
r_clean=np.char.replace(reading,'"','')
rscore=r_clean.astype("int32")
print(np.mean(rscore))
```
