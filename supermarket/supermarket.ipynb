```python
total=np.genfromtxt("SuperMarket Analysis.csv",delimiter=',',dtype=str,skip_header=1,usecols=9)
clean_total=np.char.replace(total,'"','')
tot=clean_total.astype("float")
mean=np.mean(tot)
std=np.std(tot)
print("mean= ",mean, "\n","std= ",std)
# if cv <0.2 => data consistent
# if cv>0.2-0.5 => less unstable
# if cv>0.5 => unstable inconsistent
print("coefficient of variation(cv)= ",mean/std)
```
